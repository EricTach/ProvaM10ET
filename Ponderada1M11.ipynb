# Análise Exploratória e Modelagem para Detecção de Derrame Cerebral

Neste notebook, vamos realizar a análise exploratória de um dataset de detecção de derrame cerebral e construir um modelo de classificação binária usando Keras. A abordagem será orientada a objetos e o modelo será avaliado com métricas de acurácia e F1.

## Importação das Bibliotecas Necessárias

Vamos começar importando as bibliotecas necessárias para nossa análise e modelagem.


```python
from keras.models import Sequential
from keras.layers import Dense
from keras.optimizers import Adam
from keras import metrics
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

## Modelo

Criação do modelo e a análise exploratória, utilizando orientação a objetos

```python
class StrokeDetectionModel:
    def __init__(self, data_path):
        self.data_path = data_path
        self.df = pd.read_csv(data_path)
        self._explore_data()
        self._build_model()

    def _explore_data(self):
        print("Primeiras linhas do dataset:")
        display(self.df.head())

        print("\nInformações gerais do dataset:")
        print(self.df.info())

        print("\nEstatísticas descritivas do dataset:")
        print(self.df.describe())

        print("\nDistribuição das classes:")
        print(self.df['stroke'].value_counts())

        self.features = self.df.drop('stroke', axis=1)
        self.labels = self.df['stroke']

        self.features = pd.get_dummies(self.features, drop_first=True)
        self.features = self.features.fillna(0)
        self.features = StandardScaler().fit_transform(self.features)

        self.X_train, self.X_test, self.y_train, self.y_test = train_test_split(
            self.features, self.labels, test_size=0.2, random_state=42
        )

    def _build_model(self):
        self.model = Sequential([
            Dense(16, activation='relu', input_shape=(self.X_train.shape[1],)),
            Dense(8, activation='relu'),
            Dense(1, activation='sigmoid')
        ])

        self.model.compile(
            optimizer=Adam(),
            loss='binary_crossentropy',
            metrics=[metrics.BinaryAccuracy(), metrics.Precision(), metrics.Recall()]
        )

    def train(self, epochs=50, batch_size=10):
        self.history = self.model.fit(
            self.X_train, self.y_train,
            epochs=epochs,
            batch_size=batch_size,
            validation_split=0.2
        )

    def evaluate(self):
        predictions = self.model.predict(self.X_test).flatten()
        predicted_labels = (predictions > 0.5).astype(int)

        accuracy = accuracy_score(self.y_test, predicted_labels)
        f1 = f1_score(self.y_test, predicted_labels)

        print(f"Acurácia: {accuracy:.4f}")
        print(f"Métrica F1: {f1:.4f}")
```
